```python
# Import the calculation function from area_statistics.py.
from area_statistics import summarize

# Create twelve months of small, imaginary classification maps.
months = [f"2024-{month:02d}" for month in range(1, 13)]
classes = {0: "Water", 1: "Built-up"}

# Each month has four pixels: two water and two built-up.
maps = [[0, 0, 1, 1] for _ in range(12)]

# Assume each pixel covers 100 square metres.
result = summarize(months, maps, classes, pixel_area_m2=100)

# Display the first month's results.
for row in result["rows"][:2]:
    print(
        f"{row['month']} | {row['class_name']} | "
        f"{row['percentage']:.1f}% | {row['area_m2']:.0f} square metres"
    )
```
